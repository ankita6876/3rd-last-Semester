# DL MODEL FROM SCRATCH

```python
# Model definition

model = scratch_model()

model.summary()
```

```python
# Model definition

model = scratch_model()

model = Sequential([
    Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(SIZE, SIZE, 3)),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(512, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])
```

```python
# Compile the model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Check if GPU is available

print(f"Python Platform: {platform.platform()}")
print(f"Tensor Flow Version: {tf.__version__}")
print(f"Keras Version: {keras.__version__}")
print()
print(f"Python {sys.version}")
gpu = len(tf.config.list_physical_devices('GPU'))>0
print("GPU is", "available" if gpu else "NOT AVAILABLE")
```

```python
# Define weights to handle class imblance

weight_0 = len(train_PNEUMONIA_paths) / (len(train_NORMAL_paths) + len(train_PNEUMONIA_paths))
weight_1 = len(train_NORMAL_paths) / (len(train_NORMAL_paths) + len(train_PNEUMONIA_paths))

imbalanced_weight = {0: weight_0, 1: weight_1}

print(f"Weight for label 0: {weight_0:.2f}")
print(f"Weight for label 1: {weight_1:.2f}")
```

```python
# Train the model with GPU acceleration if available

val_steps_per_epoch = val_generator.n // val_generator.batch_size
train_steps_per_epoch = train_generator.n // train_generator.batch_size

if gpu:
    with tf.device('/device:GPU:0'):
        print('Training on GPU')
        history = model.fit(train_generator,
                    steps_per_epoch = train_steps_per_epoch,
                    epochs = 10,
                    validation_data = val_generator,
                    validation_steps = val_steps_per_epoch,
                    class_weight=imbalanced_weight)
else:
    print("Training on CPU")
    history = model.fit(train_generator,
                    steps_per_epoch = train_steps_per_epoch,
                    epochs = 10,
                    validation_data = val_generator,
                    validation_steps = val_steps_per_epoch,
                    class_weight=imbalanced_weight)
```

# TRANSFER LEARNING

```python
# Define and create (if not exist) the directory for saving model weights
weights_dir = os.path.expanduser("~/.deepface/weights")
os.makedirs(weights_dir, exist_ok=True)

# Load the VGG-Face model from DeepFace
# DeepFace.build_model('VGG-Face') returns the full model, including its classification head.
# We need to extract the convolutional base for transfer learning.
vgg_full_model = DeepFace.build_model('VGG-Face')

# Identify the output of the convolutional base (before the flattening and dense layers for face recognition)
# The VGG-Face model is based on VGG16. The last pooling layer is typically named 'block5_pool'.
vgg_base_output = vgg_full_model.model.get_layer('max_pooling2d_16').output
vgg_model = Model(inputs=vgg_full_model.model.input, outputs=vgg_base_output)

# Freeze the layers of the convolutional base
for layer in vgg_model.layers:
    layer.trainable = False

# Add custom layers on top of the pre-trained convolutional base
x = vgg_model.output # This is the output of 'max_pooling2d_16'
x = Flatten()(x) # Flatten the output to feed into dense layers
x = Dense(512, activation='relu')(x)
x = Dropout(0.5)(x)
predictions = Dense(1, activation='sigmoid')(x)

# Create the final model with the new classification head
model = Model(inputs=vgg_model.input, outputs=predictions)

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model.summary()
```

```python
# Train the model with GPU acceleration if available

val_steps_per_epoch = val_generator.n // val_generator.batch_size
train_steps_per_epoch = train_generator.n // train_generator.batch_size

if gpu:
    with tf.device('/device:GPU:0'):
        print('Training on GPU')
        history = model.fit(train_generator,
                    steps_per_epoch = train_steps_per_epoch,
                    epochs = 10,
                    validation_data = val_generator,
                    validation_steps = val_steps_per_epoch,
                    class_weight=imbalanced_weight)
else:
    print("Training on CPU")
    history = model.fit(train_generator,
                    steps_per_per_epoch = train_steps_per_epoch,
                    epochs = 10,
                    validation_data = val_generator,
                    validation_steps = val_steps_per_epoch,
                    class_weight=imbalanced_weight)

# Save the model
model.save('models/pneumonia_detection_model_VGGFace.h5')
```
